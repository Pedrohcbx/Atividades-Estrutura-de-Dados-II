```
import networkx as nx

class RedeSocial:
    def __init__(self):
        self.grafo = nx.Graph()

    def adicionar_usuario(self, usuario):
        if usuario not in self.grafo:
            self.grafo.add_node(usuario)
            print(f'Usuário {usuario} adicionado.')
        else:
            print(f'Usuário {usuario} já existe.')

    def remover_usuario(self, usuario):
        if usuario in self.grafo:
            self.grafo.remove_node(usuario)
            print(f'Usuário {usuario} removido.')
        else:
            print(f'Usuário {usuario} não encontrado.')

    def adicionar_conexao(self, usuario1, usuario2):
        if usuario1 in self.grafo and usuario2 in self.grafo:
            self.grafo.add_edge(usuario1, usuario2)
            print(f'Conexão entre {usuario1} e {usuario2} adicionada.')
        else:
            print('Um ou ambos os usuários não existem.')

    def remover_conexao(self, usuario1, usuario2):
        if self.grafo.has_edge(usuario1, usuario2):
            self.grafo.remove_edge(usuario1, usuario2)
            print(f'Conexão entre {usuario1} e {usuario2} removida.')
        else:
            print('A conexão não existe.')

    def consultar_amigos(self, usuario):
        if usuario in self.grafo:
            amigos = list(self.grafo.neighbors(usuario))
            print(f'Amigos de {usuario}: {amigos}')
        else:
            print(f'Usuário {usuario} não encontrado.')

    def verificar_conexao(self, usuario1, usuario2):
        if self.grafo.has_edge(usuario1, usuario2):
            print(f'{usuario1} e {usuario2} estão conectados.')
        else:
            print(f'{usuario1} e {usuario2} não estão conectados.')

    def sugerir_conexoes(self, usuario):
        if usuario in self.grafo:
            amigos = list(self.grafo.neighbors(usuario))
            sugestoes = set()
            for amigo in amigos:
                sugestoes.update(self.grafo.neighbors(amigo))
            sugestoes.difference_update(amigos)
            sugestoes.discard(usuario)
            print(f'Sugestões de conexões para {usuario}: {sugestoes}')
        else:
            print(f'Usuário {usuario} não encontrado.')

    def grau_influencia(self, usuario):
        if usuario in self.grafo:
            grau = len(list(self.grafo.neighbors(usuario)))
            print(f'O grau de influência de {usuario} é {grau}.')
        else:
            print(f'Usuário {usuario} não encontrado.')

    def detectar_comunidades(self):
        comunidades = list(nx.connected_components(self.grafo))
        print('Comunidades detectadas:')
        for i, comunidade in enumerate(comunidades, 1):
            print(f'Com unidade {i}: {comunidade}')
```

```
rede = RedeSocial()

rede.adicionar_usuario("Alice")
rede.adicionar_usuario("Bob")
rede.adicionar_usuario("Carlos")
rede.adicionar_usuario("Diana")

rede.adicionar_conexao("Alice", "Bob")
rede.adicionar_conexao("Bob", "Carlos")
rede.adicionar_conexao("Carlos", "Diana")

rede.consultar_amigos("Bob")

rede.verificar_conexao("Alice", "Carlos")
rede.verificar_conexao("Alice", "Diana")

rede.sugerir_conexoes("Alice")

rede.grau_influencia("Bob")

rede.detectar_comunidades()
```
